```python
from google.colab import files

uploaded = files.upload()
```

```python
import os
import shutil

# Delete old frames folder if it exists
if os.path.exists("frames"):
    shutil.rmtree("frames")

# Create a new empty frames folder
os.makedirs("frames", exist_ok=True)

print("Old frames deleted. New frames folder created.")
```

```python
import cv2
import os

video_path = list(uploaded.keys())[0]

cap = cv2.VideoCapture(video_path)

fps = cap.get(cv2.CAP_PROP_FPS)

output_folder = "frames"
os.makedirs(output_folder, exist_ok=True)

frame_count = 0
saved = 0

while True:
    ret, frame = cap.read()

    if not ret:
        break

    # Save 1 frame per second
    if frame_count % int(fps) == 0:
        filename = os.path.join(output_folder, f"frame_{saved:04d}.jpg")
        cv2.imwrite(filename, frame)
        saved += 1

    frame_count += 1

cap.release()

print(f"Saved {saved} frames.")
```
